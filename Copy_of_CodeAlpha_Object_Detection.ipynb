```python
import cv2
from ultralytics import YOLO
from google.colab.patches import cv2_imshow

# Download a sample video for testing
!wget https://github.com/intel-iot-devkit/sample-videos/raw/master/person-bicycle-car-detection.mp4 -O input_video.mp4
```

```python
# Load the pre-trained YOLOv8 model (nano version is fast and lightweight)
model = YOLO('yolov8n.pt')

# Perform tracking on the video
# 'persist=True' tells the tracker to keep track of objects across frames
results = model.track(source='input_video.mp4', conf=0.3, save=True, persist=True)

print("Processing complete! Check the 'runs/detect/predict...' folder in your Colab file explorer for the output video.")
```
